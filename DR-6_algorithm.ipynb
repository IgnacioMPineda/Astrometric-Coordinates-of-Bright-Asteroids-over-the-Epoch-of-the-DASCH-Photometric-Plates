## Non-Patrol dataset
### Set-up:
The algorithm will designate two functions withing two regions that will be under the datasets, such that each sample point at x'lies above the function at y(x'). In the region to the left of the minimum point of the set of the function will be a negatively-sloped line and to teh right it will be a horizontal line.

The line will be defined by the two minimum points within that region where the minimum is P1(x1, y1) and the other point is P2(x2, y2), such that the horizontal line of the other region lies over y = y1.

### First region: Negatively-sloped line
It can be visually seen that the minimum sample of the set in regards to the average limiting magnitude is situated around x = 1960 and y = 17.5. The second point by which to define the line wil be the second minimum point within the x region by which the negatively-sloped line will be defined lies slightly beyond x = 1925 and over y = 17, these estimates also provide a visual proof that the line will be completely contained under the graphs that define the dataset, sp it will be as conservative as possible.

### Second region: Horizontal line
Just the horizontal line at y = y1= 17.5.

### Obtatining the value of the function:

```python
import numpy as np
import matplotlib.pyplot as plt
P1 = (1960, 17.50)
P2 = (1925, 17.00)

# slope
m = (P1[1] - P2[1]) / (P1[0] - P2[0])
b = P1[1] - m * P1[0]

def f(x):
    x = np.array(x)
    y = np.where(x < 1960, m * x + b, 17.5)
    return y

x = np.linspace(1900, 2000, 1000)
y = f(x)

plt.figure()  

plt.plot(x, y)
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.grid(True)
plt.xlim(1900, 2000)
plt.gca().invert_yaxis()  

plt.show()
    
```
