```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#generate fake data
M = 3
B = 2

xgrid = np.linspace(0,0.5,100)
y= M*xgrid + B
yerr = np.random.normal(0,0.5,100)
y_obs = y+yerr

plt.scatter(xgrid,y)
plt.scatter(xgrid,y_obs)
```

```python
np.polyfit(xgrid,y_obs,deg=1) #mx + b
```

```python
def error_func(M,B):
    model = M*xgrid + B #generates a guesss as a function of M,B
    error = np.sum((y_obs-model)**2)
    return error
```

```python
error_func(1,3),error_func(3,2)
```

```python
Mgrid = np.linspace(-5,5,100)
Bgrid = np.linspace(-3,3,100)
mat = []
for M in Mgrid:
    row= []
    for B in Bgrid:
        row.append(error_func(M,B))
    mat.append(row)
mat=np.array(mat)
```

```python
plt.imshow(mat,extent=[Bgrid.min(),Bgrid.max(),Mgrid.min(),Mgrid.max()],origin="lower",aspect="auto")
plt.contour(Bgrid,Mgrid,mat,colors="white",levels=50)
plt.xlabel("B")
plt.ylabel("M")
plt.scatter(2,3,c='r')
```
