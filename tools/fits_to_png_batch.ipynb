```python
from pathlib import Path
import matplotlib.pyplot as plt
from astropy.io import fits
from astropy.visualization import simple_norm

folder = Path("Data_for_reduced_images") #change this to the directory of fits files 
out = Path("pngs") #will produce an ouptu directory with pngs called 'pngs'
out.mkdir(exist_ok=True)

for f in sorted(folder.glob("*.fits")):
    data = fits.getdata(f) #load fits
    norm = simple_norm(data, "sqrt", percent=99) #sqrt normalization

    plt.figure(figsize=(8, 8))
    plt.imshow(data, origin="lower", cmap="gray", norm=norm)
    plt.title(f.stem)
    plt.colorbar()
    plt.savefig(out / f"{f.stem}.png", dpi=200, bbox_inches="tight")
    plt.close()
```
